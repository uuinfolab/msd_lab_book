## Objectives

- Refresh (and acquire if/where needed) basic skills for text and data file processing and summarisation:  libraries to handle json files, e.g. json in Python, basic text processing, e.g. regular expressions, command line programs, handling semi-structured data, databases, producing and visualising descriptive statistics, ...
- Familiarise yourself with a common data format (social media JSON API).
- Reflect on and/or identify possible data quality and validity issues.
- Generate datasets to be used during the course.
- Practice documenting social data.
