## Data description

The provided data, collected as part of the [PolarVis project](https://polarvis.github.io/) and processed into the following files for this course by the teacher, contains:

- **tweets.dat:** Tweets collected in 2023 from the period 30/11 to 12/12 2015 using the Academic API and hashtag #cop21 as a search key, without asking for consent from the monitored users. The file contains one row for each tweet. This is a monitoring of the [United Nations Conference of the Parties (COP)](https://unfccc.int/process/bodies/supreme-bodies/conference-of-the-parties-cop), where issues related to climate change are discussed and decisions are made. This is one of the historically most important COP editions, held in Paris on the same dates, leading to the [Paris Agreement](https://en.wikipedia.org/wiki/Paris_Agreement).
- **accounts.tsv:** a list of top-producers, with manual annotations about their main language, their type, and their stance. Credits to Katrin Uba and Alexandra Segerberg, Department of Political Science, who performed the qualitative annotation by manually looking at the latest tweets of each of the actors with most tweets for the COP conferences from 2015 to 2023. The language was automatically guessed using a Python library.
- **media/:** images contained in tweets posted by top producers.
- **media_list.txt:** list of image files contained in the media/ folder.

> **Note1:** The data may take time to download, even a few hours depending on the connection. So we recommend to start the download in advance. However, most of the data volume comes from the images, and for this lab you do not need to access the image files.

### Network data

Graphs with retweet and reply networks, where the vertices are accounts and the edges indicate that one account has replied to another, or retweeted a tweet from another. Think about what vertex and/or edge attributes could be useful to store.

#### Student workspace


```python
# Example code snippet
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats


fig, ax = plt.subplots(dpi=300)
x = np.linspace(0, 1, 1000)
β = 2.3
y = β * x + np.random.standard_normal(1000) * 0.3
ax.scatter(x, y, c="white", edgecolor="blue")

β_hat = np.sum(x * y) / np.sum(x**2)
y_pred = β_hat * x

ax.plot(x, y_pred, color='red', linewidth=2, label=f'OLS Fit ($\\hat{{\\beta}}$ = {β_hat:.3f})')

ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.legend()
plt.show()
```
