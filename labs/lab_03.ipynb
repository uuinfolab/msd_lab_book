## Objectives

In this lab you will keep practicing generating network data from the base data you received during the first week. This time, instead of using ties that are already present in the input data (retweets and replies), you will generate new ties between actors showing a similar behaviour. Then, while during the first lab you have computed actor-level measures (centrality) and looked at whole-network measures and function, you will now look for meso-level structures (components and communities). Finally, you will "close the circle" and map selected components and communities back to the original data, to characterise your discoveries.


## Task 1

In this first task you will generate two networks based on the URLs and/or domains posted by the accounts. Let us call *R* an entity that can be present in a tweet (*R* being a specific URL or a top-level domain). A co-action between accounts *A* and *B* occurs when *A* and *B* post one tweet each at no more than *s* seconds from each other, with the two tweets both containing the same entity *R*. A tie between *A* and *B* is generated if at least *r* co-actions have occurred between A and B. Varying the parameters above, generate two networks:

- **One aimed at identifying bots.** For example, using *s* = one second and *R* representing URLs. *You can assume that it is unlikely for two "human-driven" accounts to independently post the same URL (e.g. [http://www.lejdd.fr/Politique/Laurent-Fabius-au-JDD-Oui-j-etais-tres-emu-763913](http://www.lejdd.fr/Politique/Laurent-Fabius-au-JDD-Oui-j-etais-tres-emu-763913)) at the same time.*
- **One aimed at identifying ideologically- or topically-related groups** of accounts, for example groups of accounts posting the same URL(s) even though not necessarily at the same exact time.


### Tips & questions

- Think of whether you should consider or not retweets to generate this network.
- When you choose the filters (the type of entity *R*, the time interval s, and the number of required co-actions *r*), try to identify values leading to "reasonable" networks: if the resulting network is too dense, it may take too long to analyse it and (most importantly) the network structure may not be usable as a proxy to identify bots and/or ideology.
- Use the code you wrote last week, for example the one generating an overview of the network structure, to decide good parameter settings.
- Can you use any knowledge you have learned during the first seminar (week one) to decide which entities to use to define the ideology network?

## Reflection point

Two weeks ago you received some data with the information that you would use it to perform some social network analysis. Now you have generated various types of networks from this data - and you could keep producing many other types of derived data from it. This is quite typical in explorative studies based on secondary data. To produce the data, you have taken a number of data management decisions, such as writing an ad hoc script for each network generation task, or writing a general script converting all the information in the json file into data structures (such as Python dictionaries) and only adding specific output functions for each network, or producing intermediate data (such as CSV files, a relational database, ...) that you could then use to generate the required views. You have also made decisions about the languages to manipulate the data (All in Python? Or mostly SQL? Or have you relied on Bash commands such as AWK?) **Would you make the same choices if you had to do this again?**
